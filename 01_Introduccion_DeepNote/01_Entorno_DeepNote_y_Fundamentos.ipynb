### 📊 2. Creación y Carga de Datasets en DeepNote
DeepNote permite integrar fuentes de datos locales, URLs remotas y bases de datos SQL de manera nativa. A continuación creamos un dataset de prueba para verificar persistencia.


```python
import pandas as pd
import numpy as np

# Creamos un dataset de prueba representativo para el entorno
data = {
    'id_registro': [f"REG-{i:03d}" for i in range(1, 11)],
    'modulo': ['Intro DeepNote', 'Python Repaso', 'Librerias DS', 'Git & GitHub', 'EDA Basico'] * 2,
    'horas_dedicadas': [4.5, 6.0, 8.5, 5.0, 7.5, 4.0, 5.5, 9.0, 6.5, 8.0],
    'estado_completado': [True, True, True, True, False, True, True, True, True, False],
    'calificacion_autoevaluacion': [9.5, 8.8, 9.0, 10.0, 7.5, 8.5, 9.2, 9.8, 8.9, 7.0]
}

df_entorno = pd.DataFrame(data)
os.makedirs('data', exist_ok=True)
df_entorno.to_csv('data/deepnote_demo.csv', index=False)

print("✅ Dataset generado y guardado exitosamente en 'data/deepnote_demo.csv'")
df_entorno.head()
```

---
### 📈 3. Visualización Básica Integrada
Comprobación de la renderización gráfica dentro de celdas de DeepNote.


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.bar(df_entorno['modulo'][:5], df_entorno['horas_dedicadas'][:5], color='#2b5c8f', edgecolor='black')
plt.title("Horas Dedicadas por Módulo Temático", fontsize=12, fontweight='bold')
plt.xlabel("Módulo")
plt.ylabel("Horas")
plt.xticks(rotation=15)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
