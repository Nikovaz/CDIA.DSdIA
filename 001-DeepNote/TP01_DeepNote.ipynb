### 2. Estructuración de Celdas y Manejo de Datos en Memoria


```python
import pandas as pd
import numpy as np

# Creación de DataFrame representativo para validación de DeepNote
datos_cursada = {
    "modulo_id": [1, 2, 3, 4],
    "nombre_modulo": ["DeepNote Colaborativo", "Python Fundamentos", "Git & GitHub Versionado", "TPI - EDA"],
    "estado": ["Completado", "Completado", "En Curso", "Planificado"],
    "horas_estimadas": [4, 8, 6, 20]
}

df_deepnote = pd.DataFrame(datos_cursada)
print("📊 Resumen del Estado de Actividades en DeepNote:")
df_deepnote
```

---
### 3. Visualización Básica de Verificación


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 3.5))
plt.barh(df_deepnote["nombre_modulo"], df_deepnote["horas_estimadas"], color="#2b5c8f", edgecolor="black")
plt.xlabel("Horas Estimadas")
plt.title("Distribución de Carga Horaria por Módulo (DSIA)", fontsize=11, fontweight="bold")
plt.grid(axis="x", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
