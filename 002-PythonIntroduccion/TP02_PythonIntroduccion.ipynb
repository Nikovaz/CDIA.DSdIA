## Ejercicio 3 – Listas y bucles
Crear una lista con 5 números, imprimir cada número con un bucle `for`, y calcular suma total y promedio.

```python
numeros = [12, 45, 7, 33, 19]

print("Elementos de la lista:")
for num in numeros:
    print(num)

suma_total = sum(numeros)
promedio = suma_total / len(numeros)

print(f"\nLa suma total es: {suma_total}")
print(f"El promedio es: {promedio}")
```
