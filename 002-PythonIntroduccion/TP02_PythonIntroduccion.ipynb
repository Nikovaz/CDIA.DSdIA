### 3. Funciones Modulares y Manejo Robusto de Excepciones


```python
from typing import List, Dict, Union

def normalizar_datos(valores: List[Union[int, float]]) -> List[float]:
    """Normaliza una lista numérica al rango [0, 1] mediante Min-Max Scaling."""
    if not valores:
        raise ValueError("La lista de valores no puede estar vacía.")
    
    min_v, max_v = min(valores), max(valores)
    rango = max_v - min_v
    if rango == 0:
        return [0.0 for _ in valores]
        
    return [round((x - min_v) / rango, 4) for x in valores]

try:
    scores_muestra = [450, 720, 890, 560, 920, 610]
    scores_norm = normalizar_datos(scores_muestra)
    print(f"Valores originales: {scores_muestra}")
    print(f"Valores normalizados [0, 1]: {scores_norm}")
except ValueError as e:
    print(f"❌ Error: {e}")
```
