### 2. Estructura de Ignorados (`.gitignore`) para Proyectos de IA
Es fundamental evitar subir archivos temporales, checkpoints pesados o credenciales sensibles:
* `.ipynb_checkpoints/`
* `__pycache__/`
* `*.pyc`
* `.env` (claves API y tokens)
* `venv/` / `.venv/`
* Archivos temporales del sistema (`.DS_Store`, `Thumbs.db`)


```python
import os

def simular_verificacion_gitignore(archivos_proyecto):
    patrones_ignorados = [".ipynb_checkpoints", "__pycache__", ".env", ".venv", ".DS_Store"]
    a_trackear = []
    a_ignorar = []
    
    for arc in archivos_proyecto:
        if any(pat in arc for pat in patrones_ignorados):
            a_ignorar.append(arc)
        else:
            a_trackear.append(arc)
            
    return a_trackear, a_ignorar

archivos_simulados = [
    "001-DeepNote/TP01_DeepNote.ipynb",
    "001-DeepNote/.ipynb_checkpoints/TP01-checkpoint.ipynb",
    "002-PythonIntroduccion/TP02_PythonIntroduccion.ipynb",
    "003-GitGithub/__pycache__/utils.cpython-311.pyc",
    "003-GitGithub/TP03_GitGitHub.ipynb",
    ".env",
    "README.md"
]

track, ignore = simular_verificacion_gitignore(archivos_simulados)
print(f"✅ Archivos a versionar ({len(track)}): {track}")
print(f"🚫 Archivos ignorados ({len(ignore)}): {ignore}")
```
