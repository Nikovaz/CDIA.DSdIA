### 🧮 Parte 1: Práctica de Librerías Científicas en Python
#### 1. Creación de arreglos con NumPy

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

datos = np.array([12, 24, 36, 48, 60])
print("Array de NumPy:", datos)
```

#### 2. Manipulación y estadísticas con Pandas DataFrame

```python
# Creación del diccionario con la información
data = {
    'estudiante': ['Ana', 'Juan', 'Carla', 'Pedro', 'Lucia'],
    'nota': [8, 6, 9, 5, 7]
}

# Creación del DataFrame
df = pd.DataFrame(data)

# Mostrar el DataFrame
print("DataFrame original:")
print(df)
print("-" * 25)

# Operaciones
promedio = df['nota'].mean()
nota_maxima = df['nota'].max()
nota_minima = df['nota'].min()

print(f"Promedio de las notas: {promedio}")
print(f"Nota máxima: {nota_maxima}")
print(f"Nota mínima: {nota_minima}")
```

### 🐙 Parte 2: Control de Versiones con Git y GitHub
#### 1. Comandos Fundamentales del Flujo de Trabajo

```python
comandos_git = {
    "git init": "Inicializa el repositorio local en la raíz del proyecto.",
    "git status": "Informa el estado de los archivos (untracked, modified, staged).",
    "git add .": "Agrega los cambios al área de preparación (staging).",
    "git commit -m 'mensaje'": "Guarda una instantánea con mensaje descriptivo.",
    "git branch -M main": "Define la rama principal como 'main'.",
    "git remote add origin <url>": "Asocia el repositorio remoto de GitHub.",
    "git push -u origin main": "Sube los commits locales al repositorio remoto.",
    "git pull origin main": "Sincroniza y descarga cambios del equipo."
}

for k, v in comandos_git.items():
    print(f"📌 {k:<30} ➔ {v}")
```

#### 2. Configuración de `.gitignore` para Data Science
Para mantener limpio el repositorio en GitHub, excluimos:
- Checkpoints de notebooks (`.ipynb_checkpoints/`)
- Archivos compilados (`__pycache__/`, `*.pyc`)
- Variables de entorno y credenciales (`.env`)
- Entornos virtuales (`venv/`, `.venv/`)
