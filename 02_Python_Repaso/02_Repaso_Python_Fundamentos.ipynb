### ⚙️ 2. Funciones, Type Hinting y Manejo de Excepciones


```python
from typing import List, Dict, Union

def calcular_estadisticas_vector(valores: List[Union[int, float]]) -> Dict[str, float]:
    """
    Calcula media, mediana y desviación estándar muestral de una lista numérica.
    """
    if not valores:
        raise ValueError("La lista de valores no puede estar vacía.")
    
    n = len(valores)
    media = sum(valores) / n
    varianza = sum((x - media) ** 2 for x in valores) / (n - 1) if n > 1 else 0.0
    desvio = varianza ** 0.5
    
    valores_ordenados = sorted(valores)
    if n % 2 == 1:
        mediana = valores_ordenados[n // 2]
    else:
        mediana = (valores_ordenados[(n // 2) - 1] + valores_ordenados[n // 2]) / 2.0
        
    return {
        "conteo": float(n),
        "media": round(media, 4),
        "mediana": round(mediana, 4),
        "desvio_estandar": round(desvio, 4)
    }

# Prueba con manejo de excepciones
try:
    sample = [12.5, 14.8, 11.2, 19.4, 25.1, 14.8, 16.0]
    res = calcular_estadisticas_vector(sample)
    print("📊 Resultados estadísticos calculados:")
    for k, v in res.items():
        print(f"  • {k}: {v}")
except Exception as e:
    print(f"❌ Error durante el cálculo: {e}")
```

### 🏗️ 3. Programación Orientada a Objetos: Pipeline de Datos


```python
class DataProcessor:
    """Clase para preprocesamiento y limpieza de datos."""
    
    def __init__(self, data: List[Dict[str, any]]):
        self.data = data
        self.log = []
        
    def filtrar_nulos(self, campo_clave: str):
        prev_len = len(self.data)
        self.data = [d for d in self.data if d.get(campo_clave) is not None]
        eliminados = prev_len - len(self.data)
        self.log.append(f"Filtro nulos en '{campo_clave}': {eliminados} registros eliminados.")
        return self
        
    def normalizar_min_max(self, campo_numerico: str):
        valores = [d[campo_numerico] for d in self.data if isinstance(d.get(campo_numerico), (int, float))]
        if not valores:
            return self
        min_v, max_v = min(valores), max(valores)
        rango = (max_v - min_v) if (max_v - min_v) != 0 else 1.0
        
        for d in self.data:
            if isinstance(d.get(campo_numerico), (int, float)):
                d[f"{campo_numerico}_norm"] = round((d[campo_numerico] - min_v) / rango, 4)
                
        self.log.append(f"Normalización Min-Max aplicada a '{campo_numerico}' (Rango [{min_v}, {max_v}]).")
        return self

# Prueba del pipeline
registros_raw = [
    {"id": 1, "edad": 25, "score": 850},
    {"id": 2, "edad": None, "score": 620},
    {"id": 3, "edad": 42, "score": 910},
    {"id": 4, "edad": 31, "score": 740}
]

pipeline = DataProcessor(registros_raw).filtrar_nulos("edad").normalizar_min_max("score")
print("✅ Registros transformados:")
for r in pipeline.data:
    print(" ", r)
print("\n📜 Logs del Pipeline:")
for l in pipeline.log:
    print(" ", l)
```
