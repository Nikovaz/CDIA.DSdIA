## Ejercicio 2 – Dataset simple con Pandas
Crear un DataFrame con la información de estudiantes y notas, mostrar el DataFrame, calcular el promedio de notas, nota máxima y nota mínima.

```python
# Creación del diccionario con la información
data = {
    'estudiante': ['Ana', 'Juan', 'Carla', 'Pedro', 'Lucia'],
    'nota': [8, 6, 9, 5, 7]
}

# Creación del DataFrame
df = pd.DataFrame(data)

# Mostrar el DataFrame
print("DataFrame original:")
print(df)
print("-" * 25)

# Operaciones
promedio = df['nota'].mean()
nota_maxima = df['nota'].max()
nota_minima = df['nota'].min()

print(f"Promedio de las notas: {promedio}")
print(f"Nota máxima: {nota_maxima}")
print(f"Nota mínima: {nota_minima}")
```
